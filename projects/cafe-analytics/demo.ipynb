```python
# read bigquery data into pandas dataframe
import pandas as pd
df = pd.read_gbq(
    """
    SELECT  count(*)
    FROM `jr-data-dev training.dbt_cafeanalytics.fact_product_sales`
  """,
    project_id="jr-data-training",
    location="australia-southeast1",
)
```
